```python
reservoirs=[]
for block in range(1,4):
  print(f"....Block {block}....")
  area=float(input("Area m2: "))
  thickness=float(input("Thickness m:"))
  porosity=float(input("Porosity 0-1 :"))
  oil= area * thickness * porosity
  reservoirs.append(oil)
print(f"\nAll blocks:{reservoirs}")
print(f"Biggest block has: {max(reservoirs)}")
biggest= max(reservoirs)
block_num=reservoirs.index(biggest) + 1
print(f"Biggest block has: {round(biggest, 2 )} m³")
print(f"block {block_num}"
```
